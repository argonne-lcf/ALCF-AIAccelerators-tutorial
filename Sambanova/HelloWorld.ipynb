## Let's define a few variables up here so they are easier to find and change.

```python
model = "DeepSeek-V3.2"  # This can be any model available to you in "Models Available Today" SambaNova Cloud (https://cloud.sambanova.ai/dashboard)

temerature = 1.2
# In the context of large language models (LLMs), "model temperature" is a parameter that controls the randomness or predictability of the 
# model's output during text generation. Think of it as a dial that adjusts how much the model "risks" or "safely" generates text.  

# * Lower temperature (e.g., 0.2):
# The model becomes more predictable and sticks to the most likely next words, resulting in more focused and deterministic outputs. This
# is useful for tasks requiring accuracy and precision, like question answering or technical writing.

# * Higher temperature (e.g., 1.0 or 1.2):
# The model takes more risks and generates more diverse and creative outputs, potentially including less likely or even unconventional 
# words. This can be useful for tasks like storytelling, poetry, or brainstorming.

# * Temperature of 1.0 (or the default):
# Often provides a good balance between creativity and logical consistency, suitable for general-purpose text generation or conversational AI. 

top_p=0.1

# In large language models (LLMs), top_p, also known as nucleus sampling, is a parameter that controls the diversity of the generated text. 
# It dictates which tokens (words or parts of words) are considered when the model predicts the next word. By adjusting top_p, you can influence 
# the model's tendency to choose from a limited set of most likely words or to explore a wider range of possibilities, including less common words. 

# * Low top_p (e.g., 0.2): The model focuses on the most likely words, leading to more predictable and coherent text, but potentially less diverse output. 
# * High top_p (e.g., 0.9): The model considers a wider range of tokens, including less likely ones, resulting in more creative and diverse output,
# but possibly with some loss of coherence. 

# * Value range: top_p values typically range from 0.0 to 1.0. 
```

## Now lets define some variables we are going to play with

```python
Persona_1 = "You are a very friendly Artificial Intelligence"
Persona_2 = "You are a representative speaking for all of the worlds people and you are very nervious about AI"

start_prompt = "Provide your very first greeting to the entire planet in only one sentance"
```

```python
def create_completion(client, model, system_content, prompt, temperature, top_p):
    
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "system", "content": system_content}, {"role": "user", "content": prompt}],
        temperature=temperature,
        top_p=top_p
    )
    return response
```

```python
client = OpenAI(
    base_url="https://api.sambanova.ai/v1/",
    api_key=my_key,
)

prompt = start_prompt
response1 = create_completion(client,model,Persona_1,prompt,temerature,top_p).choices[0].message.content
print(response1)
```
